# Data Preprocessing

```python
import nltk
from nltk.corpus import stopwords
nltk.download("stopwords")
```

```python
import regex as re
stop_words=set(stopwords.words("english"))

def text_preprocessing(text):
    text=text.lower()

    text = re.sub(r'\bx+\b', '', text)
    
    text = re.sub(r'[$₹€£]\s?[\d,]+(?:\.\d+)?', ' money ', text)
    
    text=re.sub(r'[^a-z0-9\s]','',text)
    
    tokens=text.split()

    preprocessed_text=[w for w in tokens if w not in stop_words]

    return ' '.join(preprocessed_text)
```

```python
df["Consumer complaint narrative"]=df["Consumer complaint narrative"].apply(text_preprocessing)
```

# creating embedding matrix

```python
embedding_dim=100
vocab_size=40000

embedding_matrix=np.zeros((vocab_size,embedding_dim))

word_index=tokenizer.word_index
word_index

for word,index in word_index.items():
    if index>=vocab_size:
        continue

    if word in fasttext.wv:
        embedding_matrix[index]=(fasttext.wv[word])
```

# Model Building

```python
from keras.models import Sequential,Model

from keras.layers import Input,Embedding,Dense,Dropout,Bidirectional,LSTM
```

```python
num_classes = len(np.unique(y_problem_train))
num_classes
```

```python
inputs = Input(shape=(x_train_pad.shape[1],))

X = Embedding(input_dim=vocab_size,output_dim=embedding_dim, weights=[embedding_matrix],trainable=False)(inputs)

x1 = Bidirectional(LSTM(128))(X)
x1 = Dropout(0.30)(x1)

x2 = Dense(128,activation="relu")(x1)
x2 = Dropout(0.30)(x2)

x3 = Dense(64,activation="relu")(x2)
x3 = Dropout(0.20)(x3)


Problem_output = Dense(12,activation="softmax",name="problem")(x3)


model = Model(inputs=inputs,outputs=[Problem_output])
model.summary()
```

```python
import tensorflow as tf
```

```python
model.compile(optimizer="adam",loss=["sparse_categorical_crossentropy"],metrics=["accuracy"])
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
```

```python
problem_class_weights= compute_class_weight(class_weight="balanced",y=y_problem_train,classes=np.unique(y_problem_train))
print(problem_class_weights)
```

```python
classes=np.unique(y_problem_train)
problem_class_weights = dict(zip(classes, problem_class_weights))
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
early_stop = EarlyStopping(monitor="accuracy",patience=10,min_delta=0.0001,restore_best_weights=True)
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint
checkpoint = ModelCheckpoint(
    filepath='best_model.keras',  # Where to save the file
    monitor='val_loss',  # Metric to watch
    mode='min',  # 'max' for accuracy, 'min' for loss
    save_best_only=True,  # Only save when the model improves
    verbose=1,
)
```

```python
from sklearn.metrics import f1_score

class F1Callback(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        pred = np.argmax(self.model.predict(x_test_pad, verbose=0), axis=1)
        f1 = f1_score(y_problem_test, pred, average="macro")
        print("F1 Score:", f1)
```

```python
model.fit(x_train_pad,y_problem_train,validation_data=(x_test_pad,y_problem_test),class_weight=problem_class_weights,callbacks=[early_stop,checkpoint,F1Callback()],epochs=20,batch_size=50)
```

# Bayesian Optuna

import optuna 
from keras.layers import LeakyReLU,PReLU
from keras.optimizers import SGD,Adagrad,RMSprop,Adam
from keras.layers import BatchNormalization

def objective(trial):

    lstm_units = trial.suggest_int('lstm_units', 64, 256)

    n_layers = trial.suggest_int('n_layers', 1, 10)

    epochs = trial.suggest_int('epochs', 1, 100)

    learn_rate = trial.suggest_float('learning_rate', 1e-4, 1e-2)

    batch_size = trial.suggest_int('batch_size', 6, 128)

    activation = trial.suggest_categorical('activation',["sigmoid", "tanh", "relu"])

    optimizers = trial.suggest_categorical('optimizers',['SGD', 'Adagrad', 'RMSprop', 'Adam'])

    dropout1 = trial.suggest_float('dropout1', 0.1, 0.5)


    inputs = Input(shape=(x_train_pad.shape[1],))


    X = Embedding(input_dim=vocab_size,output_dim=embedding_dim,weights=[embedding_matrix],trainable=False)(inputs)


    x1 = Bidirectional(LSTM(lstm_units))(X)

    x1 = Dropout(dropout1)(x1)


    x3 = x1

    for i in range(n_layers):

        n_units = trial.suggest_int(f'units{i}',32,128)

        x3 = Dense(units=n_units,activation=activation)(x3)


    Problem_output = Dense(5,activation="softmax",name="problem")(x3)


    Issue_output = Dense(98,activation="softmax",name="issue")(x3)


    SubIssue_output = Dense(213,activation="softmax",name="subissue")(x3)


    model = Model(inputs=inputs,outputs=[Problem_output,Issue_output,SubIssue_output])


    if optimizers == 'SGD':

        optimizer = SGD(learning_rate=learn_rate)

    elif optimizers == 'Adagrad':

        optimizer = Adagrad(learning_rate=learn_rate)

    elif optimizers == 'RMSprop':

        optimizer = RMSprop(learning_rate=learn_rate)

    else:

        optimizer = Adam(learning_rate=learn_rate)


    model.compile(optimizer=optimizer,loss=['sparse_categorical_crossentropy','sparse_categorical_crossentropy',
            'sparse_categorical_crossentropy'],
        metrics=['accuracy','accuracy','accuracy'])


    history = model.fit(x_train_pad,[y_problem_train,y_issue_train,y_subissue_train],
            epochs=epochs,batch_size=batch_size,
            validation_data=(x_test_pad,[y_problem_test,y_issue_test,y_subissue_test]),verbose=0)


    problem_accuracy = max(history.history['val_problem_accuracy'])

    issue_accuracy = max(history.history['val_issue_accuracy'])

    subissue_accuracy = max(history.history['val_subissue_accuracy'])


    accuracy = (problem_accuracy +issue_accuracy +subissue_accuracy) / 3


    return accuracy

study = optuna.create_study(direction='maximize')

study.optimize( objective, n_trials=20)

print(study.best_params)

print(study.best_value)
